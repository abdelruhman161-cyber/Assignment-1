## 5. Why ML beats a fixed rule here
Fixed if-else rules fail because search performance depends on complex, non-linear interactions between multiple features (e.g., position tier, CTR drop rate, word count, and intent type). A simple rule like if impressions drop > 20% generates too many false positives for high-traffic pages and misses subtle opportunities on page-2 rankings. ML captures these non-linear interactions automatically.
